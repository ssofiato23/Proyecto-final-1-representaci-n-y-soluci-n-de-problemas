# <FONT SIZE=5 COLOR="red"> **5. Represetación del problema** </FONT>

El problema se representa mediante un sistema multiagente compuesto por empresas, consumidores y un regulador que interactúan dentro de un mercado digital simulado.

## <FONT SIZE=5 COLOR="red"> **5.1 Agente Empresa** </FONT>

Las empresas son agentes autónomos encargados de fijar precios con el objetivo de maximizar sus ganancias mediante aprendizaje por refuerzo. Cada empresa posee las siguientes características:

* Precio actual del producto o servicio.
* Confiabilidad percibida por los consumidores.
* Número de ventas realizadas en cada iteración.
* Ganancia obtenida.
* Ganancia acumulada durante toda la simulación.
* Participación de mercado.
* Tabla Q utilizada para almacenar la experiencia adquirida durante el proceso de aprendizaje.

Para representar el estado del entorno, cada empresa observa:

* Su precio actual.
* El precio promedio de las empresas competidoras.
* El número de ventas obtenidas.

El estado se representa mediante la tupla:

(Precio propio, Precio promedio de la competencia, Nivel de ventas)

Con el fin de reducir el espacio de estados, estas variables se discretizan en intervalos.

Las acciones disponibles para cada empresa son:

* Disminuir el precio en una unidad (-1).
* Mantener el precio actual (0).
* Aumentar el precio en una unidad (+1).

La selección de acciones se realiza mediante una estrategia ε-greedy, que combina exploración y explotación de experiencias previas.

La recompensa utilizada durante el aprendizaje corresponde a la ganancia obtenida por la empresa después de interactuar con los consumidores. Dicha recompensa se calcula como:

Ganancia = (Precio − Costo de producción) × Ventas

Posteriormente, la tabla Q se actualiza utilizando el algoritmo Q-Learning, permitiendo que cada empresa aprenda estrategias de fijación de precios cada vez más rentables.


```python
class AgenteEmpresa(Agent):

    def __init__(
        self,
        model,
        unique_id,
        precio_inicial=25,
        confiabilidad=None,
        epsilon=0.1,
        alpha=0.2,
        gamma=0.9
    ):

        super().__init__(model)

        self.unique_id = unique_id

        self.precio = precio_inicial
        self.participacion = 0
        self.confiabilidad = (
            confiabilidad
            if confiabilidad is not None
            else random.uniform(0.6, 1.0)
        )
        self.ganancia_acumulada = 0

        self.ventas = 0
        self.ganancia = 0

        self.epsilon = epsilon
        self.alpha = alpha
        self.gamma = gamma

        self.q_values = {}

    def obtener_estado(self):

        precios_competencia = [
            empresa.precio
            for empresa in self.model.agentes_empresas
            if empresa != self
        ]

        precio_promedio = (
            sum(precios_competencia)
            / len(precios_competencia)
        )

        return (
          round(self.precio / 5),
          round(precio_promedio / 5),
          round(self.ventas / 10)
        )

    def elegir_accion(self):

        acciones = [-1, 0, 1]

        estado = self.obtener_estado()

        if random.random() < self.epsilon:
            return random.choice(acciones)

        valores = [
            self.q_values.get(
                (estado, accion),
                0
            )
            for accion in acciones
        ]

        mejor_valor = max(valores)

        mejores_acciones = [
            accion
            for accion, valor in zip(
                acciones,
                valores
            )
            if valor == mejor_valor
        ]

        return random.choice(
            mejores_acciones
        )

    def actualizar_q(
        self,
        estado,
        accion,
        recompensa,
        nuevo_estado
    ):

        valor_actual = self.q_values.get(
            (estado, accion),
            0
        )

        acciones = [-1, 0, 1]

        mejor_futuro = max(
            [
                self.q_values.get(
                    (nuevo_estado, a),
                    0
                )
                for a in acciones
            ]
        )

        nuevo_valor = (
            valor_actual
            + self.alpha
            * (
                recompensa
                + self.gamma
                * mejor_futuro
                - valor_actual
            )
        )

        self.q_values[
            (estado, accion)
        ] = nuevo_valor

    def step(self):

        pass

    def calcular_ganancia(self):

        self.ganancia = (
            (self.precio -
            self.model.costo_produccion)
            * self.ventas
        )

        self.ganancia_acumulada += self.ganancia

        return self.ganancia
```

## <FONT SIZE=5 COLOR="red"> **5.2 Agente Consumidor** </FONT>

Los consumidores representan la demanda del mercado y son los encargados de decidir a qué empresa realizar una compra en cada iteración.

Cada consumidor posee preferencias individuales que determinan la importancia relativa que asigna al precio y a la confiabilidad de las empresas. Estas preferencias se representan mediante pesos aleatorios, permitiendo modelar consumidores heterogéneos con diferentes criterios de decisión.

Para cada empresa, el consumidor calcula una utilidad basada en dos factores:

* Confiabilidad ofrecida por la empresa.
* Precio del producto o servicio.

La utilidad se calcula mediante una combinación ponderada de ambos criterios:

Utilidad = (Peso de confiabilidad × Confiabilidad) − (Peso de precio × Precio)

Una vez calculadas las utilidades, estas se utilizan para construir una distribución de probabilidades sobre las empresas disponibles. De esta forma, las empresas con mayor utilidad tienen una mayor probabilidad de ser seleccionadas, aunque no necesariamente son elegidas en todos los casos.

Este mecanismo permite representar la toma de decisiones bajo incertidumbre e incorpora elementos de inferencia probabilística dentro del sistema multiagente, produciendo una distribución más realista de la demanda entre las empresas.

Cuando un consumidor selecciona una empresa, se incrementa en una unidad el número de ventas de dicha empresa.


```python
class AgenteConsumidor(Agent):

    def __init__(
        self,
        model,
        unique_id
    ):

        super().__init__(model)

        self.unique_id = unique_id

        self.peso_precio = random.uniform(
            0.4,
            0.8
        )

        self.peso_confiabilidad = (
            1 - self.peso_precio
        )

    def elegir_empresa(self):

        utilidades = []

        for empresa in self.model.agentes_empresas:

            utilidad = (
                self.peso_confiabilidad
                * empresa.confiabilidad
                -
                self.peso_precio
                * (empresa.precio / 100)
            )

            utilidad = max(
                utilidad,
                0.01
            )

            utilidades.append(utilidad)

        empresa_elegida = random.choices(
            self.model.agentes_empresas,
            weights=utilidades,
            k=1
        )[0]

        return empresa_elegida
    def step(self):

        empresa = self.elegir_empresa()

        empresa.ventas += 1
```

## <FONT SIZE=5 COLOR="red"> **5.3 Agente Regulador** </FONT>

El regulador representa una entidad encargada de supervisar el comportamiento general del mercado e identificar posibles indicios de colusión entre las empresas.

Para ello, monitorea dos indicadores principales:
- Convergencia de precios, medida como la diferencia entre el precio máximo y el precio mínimo de las empresas.
- Índice de Lerner, utilizado para estimar el poder de mercado existente en cada iteración.

El Índice de Lerner se calcula como:
$$
L=
\frac{P -C }{P}
$$
donde P corresponde al precio promedio del mercado y C al costo de producción.

El regulador considera que existe un posible comportamiento colusivo cuando se cumplen simultáneamente las siguientes condiciones:
- La diferencia entre el precio máximo y el precio mínimo es menor o igual a 5 unidades.
- El Índice de Lerner es superior a 0.3.

Cuando estas condiciones se satisfacen, el regulador genera una alerta y registra un posible episodio de colusión. En caso contrario, considera que el mercado continúa operando bajo condiciones normales de competencia.

La principal función de este agente es proporcionar un mecanismo de monitoreo que permita evaluar si las estrategias aprendidas por las empresas conducen a comportamientos compatibles con una posible coordinación de precios.

```python
class AgenteRegulador(Agent):

    def __init__(
        self,
        model,
        unique_id
    ):

        super().__init__(model)

        self.unique_id = unique_id

        self.numero_alertas = 0

        self.colusion_detectada = False

    def calcular_lerner(self):

        precios = [
            e.precio
            for e in self.model.agentes_empresas
        ]

        precio_promedio = (
            sum(precios)
            / len(precios)
        )

        return (
            precio_promedio
            - self.model.costo_produccion
        ) / precio_promedio

    def detectar_colusion(self):

        precios = [
            e.precio
            for e in self.model.agentes_empresas
        ]

        diferencia = (
            max(precios)
            - min(precios)
        )

        lerner = (
            self.calcular_lerner()
        )

        if (
            diferencia <= 5
            and lerner > 0.3
        ):

            self.colusion_detectada = True

            self.numero_alertas += 1

        else:

            self.colusion_detectada = False

    def step(self):

        self.detectar_colusion()
```

## <FONT SIZE=5 COLOR="red"> **5.4 Modelo** </FONT>

El modelo desarrollado representa un mercado digital compuesto por tres tipos de agentes: empresas, consumidores y un regulador. El número de empresas y consumidores puede configurarse al inicio de la simulación, permitiendo analizar diferentes escenarios de mercado. Cada empresa inicia con un precio aleatorio dentro de un rango definido, mientras que los consumidores poseen preferencias individuales respecto al precio y la confiabilidad de las empresas. Adicionalmente, se incorpora un agente regulador encargado de supervisar el comportamiento general del mercado.

Las empresas utilizan aprendizaje por refuerzo mediante Q-Learning para ajustar sus precios con el objetivo de maximizar sus ganancias. Para ello, observan el estado del mercado, seleccionan una acción y posteriormente actualizan su conocimiento a partir de la recompensa obtenida.

Los consumidores representan la demanda del mercado y, en cada iteración, evalúan las empresas disponibles utilizando una función de utilidad basada en el precio y la confiabilidad. A partir de estas utilidades se construye una distribución de probabilidades que determina la empresa seleccionada para realizar la compra. Este mecanismo permite modelar decisiones bajo incertidumbre y generar una distribución más realista de la demanda.

Por su parte, el regulador monitorea la evolución de los precios y calcula indicadores que permiten identificar posibles comportamientos coordinados dentro del mercado.

La simulación se ejecuta de forma iterativa siguiendo la siguiente secuencia:

1. Se reinician las ventas de todas las empresas.
2. Las empresas observan el estado del mercado y ajustan sus precios.
3. Los consumidores evalúan las empresas y realizan una compra.
4. Se calcula la participación de mercado de cada empresa.
5. Se calculan las ganancias obtenidas por cada empresa.
6. Las empresas actualizan sus valores Q a partir de la recompensa recibida.
7. El regulador analiza el comportamiento del mercado.
8. Se almacenan las métricas de seguimiento para su posterior análisis.

Con el fin de evaluar la evolución del sistema a lo largo del tiempo, el modelo incorpora un mecanismo de recolección de datos mediante DataCollector. Este componente registra automáticamente indicadores relevantes del mercado, tales como el precio promedio, la ganancia promedio, el índice de Lerner, la convergencia de precios, la participación máxima de mercado y el número de alertas generadas por el regulador.



```python
class ModeloColusion(Model):

    def __init__(
        self,
        numero_empresas=5,
        numero_consumidores=100,
        costo_produccion=20
    ):

        super().__init__()

        self.costo_produccion = costo_produccion
        self.numero_consumidores = numero_consumidores

        self.agentes_empresas = []
        self.agentes_consumidores = []

        # -------------------
        # Empresas
        # -------------------

        for i in range(numero_empresas):

            empresa = AgenteEmpresa(
                self,
                unique_id=i,
                precio_inicial=random.randint(20, 30)
            )

            self.agentes_empresas.append(
                empresa
            )

        # -------------------
        # Consumidores
        # -------------------

        for i in range(numero_consumidores):

            consumidor = AgenteConsumidor(
                self,
                unique_id=1000 + i
            )

            self.agentes_consumidores.append(
                consumidor
            )

        # -------------------
        # Regulador
        # -------------------

        self.regulador = AgenteRegulador(
            self,
            unique_id=9999
        )

        # -------------------
        # Data Collector
        # -------------------

        self.datacollector = DataCollector(

            model_reporters={

                "PrecioPromedio":
                    lambda m: m.precio_promedio(),

                "GananciaPromedio":
                    lambda m: m.ganancia_promedio(),

                "IndiceLerner":
                    lambda m: m.indice_lerner(),

                "Convergencia":
                    lambda m: m.convergencia(),
                "ParticipacionMaxima":
                    lambda m: m.participacion_maxima(),

                "Alertas":
                    lambda m:
                    m.regulador.numero_alertas
            }
        )

    # =================================
    # MÉTRICAS
    # =================================

    def precio_promedio(self):

        return sum(
            empresa.precio
            for empresa in self.agentes_empresas
        ) / len(self.agentes_empresas)

    def ganancia_promedio(self):

        return sum(
            empresa.ganancia
            for empresa in self.agentes_empresas
        ) / len(self.agentes_empresas)

    def indice_lerner(self):

        precio = self.precio_promedio()

        return (
            precio -
            self.costo_produccion
        ) / precio

    def participacion_maxima(self):
        return max(
          empresa.participacion
          for empresa in self.agentes_empresas
        )

    def convergencia(self):

        precios = [
            empresa.precio
            for empresa in self.agentes_empresas
        ]

        return (
            max(precios)
            - min(precios)
        )

    # =================================
    # SIMULACIÓN
    # =================================

    def step(self):

        # ------------------
        # Reiniciar ventas
        # ------------------

        for empresa in self.agentes_empresas:

            empresa.ventas = 0

        # ------------------
        # Empresas fijan precio
        # ------------------

        for empresa in self.agentes_empresas:

            estado = empresa.obtener_estado()

            accion = empresa.elegir_accion()

            empresa.estado_anterior = estado
            empresa.accion_anterior = accion

            empresa.precio += accion

            empresa.precio = max(
                self.costo_produccion,
                empresa.precio
            )

        # ------------------
        # Consumidores compran
        # ------------------

        for consumidor in self.agentes_consumidores:

            consumidor.step()

        # ------------------
        # Participación de mercado
        # ------------------

        for empresa in self.agentes_empresas:

            empresa.participacion = (
                empresa.ventas
                / self.numero_consumidores
            )

        # ------------------
        # Empresas calculan
        # ganancias y aprenden
        # ------------------

        for empresa in self.agentes_empresas:

            recompensa = empresa.calcular_ganancia()

            nuevo_estado = (
                empresa.obtener_estado()
            )

            empresa.actualizar_q(
                empresa.estado_anterior,
                empresa.accion_anterior,
                recompensa,
                nuevo_estado
            )

        # ------------------
        # Regulador
        # ------------------

        self.regulador.step()

        # ------------------
        # Guardar datos
        # ------------------

        self.datacollector.collect(self)
```

# <FONT SIZE=5 COLOR="red"> **6. Variables de seguimiento** </FONT>

Para analizar el comportamiento del sistema y evaluar la evolución de las estrategias de los agentes, se definieron las siguientes variables de seguimiento. Estas métricas son registradas durante cada iteración de la simulación mediante el mecanismo de recolección de datos implementado en el modelo.

## <FONT SIZE=5 COLOR="red"> **6.1 Precio promedio** </FONT>

Corresponde al promedio de los precios establecidos por todas las empresas en una determinada iteración.

Esta variable permite analizar la evolución general del mercado y observar cómo las decisiones de aprendizaje de los agentes afectan los niveles de precio a lo largo del tiempo.

## <FONT SIZE=5 COLOR="red"> **6.2 Ganancia promedio** </FONT>

Representa el promedio de las ganancias obtenidas por las empresas durante cada iteración.

Su propósito es evaluar el desempeño de las estrategias aprendidas y determinar si los agentes logran maximizar sus beneficios mediante la adaptación continua de sus precios.

## <FONT SIZE=5 COLOR="red"> **6.3 Índice de Lerner** </FONT>

El índice de Lerner es una medida utilizada para estimar el poder de mercado de las empresas. Se calcula a partir de la diferencia entre el precio promedio del mercado y el costo de producción

$$
L=
\frac{P -C }{P}
$$
dond

donde:

* (P) representa el precio promedio del mercado.
* (C) representa el costo de producción.

Valores más altos indican una mayor diferencia entre los precios y los costos, mientras que valores bajos reflejan una mayor presión competitiva.

## <FONT SIZE=5 COLOR="red"> **6.4 Convergencia de precios** </FONT>

La convergencia de precios se calcula como la diferencia entre el precio máximo y el precio mínimo registrado entre las empresas en una misma iteración.

Esta variable permite medir el grado de similitud entre las estrategias de precios adoptadas por los agentes. Valores cercanos a cero indican una mayor convergencia entre las empresas, mientras que valores elevados reflejan una mayor diversidad de estrategias.

## <FONT SIZE=5 COLOR="red"> **6.5 Participación Máxima de Mercado** </FONT>

Corresponde a la mayor participación de mercado observada entre todas las empresas en una determinada iteración.

Esta métrica permite identificar si una empresa concentra una proporción significativa de las ventas totales del mercado. Valores elevados pueden indicar una posición dominante, mientras que valores bajos sugieren una distribución más equilibrada de la demanda entre las empresas.

## <FONT SIZE=5 COLOR="red"> **6.6 Número de Alertas** </FONT>

Corresponde a la cantidad acumulada de alertas generadas por el agente regulador durante la simulación.

Una alerta se activa cuando el Índice de Lerner supera el umbral establecido y la diferencia entre los precios de las empresas es reducida. Esta métrica permite identificar situaciones en las que el comportamiento observado en el mercado presenta características potencialmente coordinadas o compatibles con escenarios de colusión.

# <FONT SIZE=5 COLOR="red"> **7. Configuración experimental** </FONT>
